**[Task 2.1.]()** Implement Otsu thresholding as a re-usable function:

```python
def otsu_thresholding(imgf):
    img8 = (imgf * 255).round().astype(np.uint8)
    
    h = np.zeros(256, dtype=int)
    for i in range(256):
        h[i] = np.sum(img8 == i)

    var = np.zeros(256)

    var[0] = np.inf

    for T in range(1, 256) :
        h1 = h[0:T]
        h2 = h[T:256]

        n1 = np.sum(h1)
        n2 = np.sum(h2)

        var1 = np.var(img8[img8 <T])
        var2 = np.var(img8[img8 >= T])

        var[T] = n1 * var1 + n2 * var2

        T_otsu = np.nanargmin(var) 
        seg = img8 >= T_otsu
    
    return seg
```

**[Task 2.2.]()** Implement a re-usable function to compute the Dice coefficient:

```python
def compute_dice(bin1, bin2):
    assert bin1.dtype == bool
    assert bin2.dtype == bool
    
    dice = 2* (bin1 * bin2).sum() / (bin1.sum() + bin2.sum())
    return dice
```

```python
#liste erstellen
dice_scores = []

# die i's sind einfach unterschiedliche Bilder und wir berechnen für diese Bilder jetzt den dice score
for i in [28, 29, 33, 44, 46, 49]: 
    imgf = plt.imread(f'data/NIH3T3/im/dna-{i}.png')
    gt = plt.imread(f'data/NIH3T3/gt/{i}.png').astype(bool)
    seg = otsu_thresholding(imgf)
    dice = compute_dice(seg, gt)
    dice_scores.append(dice)

print(dice_scores)
dice_scores_mean = np.mean(dice_scores)
print(dice_scores_mean)
```
